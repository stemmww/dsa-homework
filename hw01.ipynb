````python
#1
cities_clean = []
for city in cities:
  city = city.strip().title()
  cities_clean.append(city)
print(cities_clean)
````

````python
#2
amounts_int = []
for amount in amounts:
  amount = int(amount)
  amounts_int.append(amount)
print(amounts_int)
````

````python
#3
total_orders = len(amounts_int)
total_amount = sum(amounts_int)
average_check = round(total_amount/total_orders)
print(total_orders)
print(total_amount)
print(average_check)
````

````python
#4
unpaid_orders = 0
unpaid_amount = 0
for i in paid:
  if i == False:
    unpaid_orders +=1
    unpaid_amount += amounts_int[paid.index(i)]
print(f"Unpaid Orders: {unpaid_orders}")
print(f"Unpaid Amount: {unpaid_amount}")
````

````python
#5
zipped = zip(cities_clean, amounts_int, paid)
for city, amounts_int, paid in zipped:
  print(f"{city} | {amounts_int} | {paid}")
````
